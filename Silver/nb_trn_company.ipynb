#Company Table Cleaning


##importing dependency and data

```python
import pyspark.sql.functions as F
```

```python
df_company_raw = spark.read.table("bronze.azure_blob_storage.company")
```

##Removing columns inserted by fivetron

```python
df_company=df_company_raw.drop('_file','_line','_modified','_fivetran_synced')
```

```python
df_company.display()
```

##Trimming column names


```python
for column in df_company.columns:
    df_company=df_company.withColumn(column,F.trim(F.col(column)))
df_company.display()
```

## Type Casting

```python
df_company.schema['is_active'].dataType
```

```python
df_company = df_company.withColumn('is_active', F.when(F.col('is_active') == "true", True).otherwise(False))
df_company.display()
```

```python
df_company  = df_company.withColumn(
    "established_date",
    F.to_date(F.col("established_date"), "dd-MM-yyyy")
)
```

```python
df_company = df_company.withColumn('established_date', F.date_format(F.col('established_date'), 'yyyy-MM-dd'))
```

```python
df_company.display()
```

```python
df_company  = df_company.withColumn(
    "established_date",
    F.to_date(F.col("established_date"), "yyyy-MM-dd")
)
```

```python
df_company.display()
```

```python
df_company.write.mode('overwrite').saveAsTable("silver.transformation.company")
```
