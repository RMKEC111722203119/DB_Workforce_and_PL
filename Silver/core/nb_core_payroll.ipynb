#Core Buisness Logic for Payroll

## Import data

```python

import pyspark.sql.functions as F
```

```python
df_silver_payroll=spark.table("silver.transformation.payroll")
```

```python
df_silver_payroll.display().limit(5)
```

## total compensation and dedection columns created

```python
df_silver_payroll = df_silver_payroll.withColumn(
    "total_compensation",
    F.col("gross_salary") + F.col("bonus") + F.col("overtime_pay") + F.col("commission") + F.col("allowances")
).withColumn(
    "total_deduction",
    F.col("tax_deduction") + F.col("social_security") + F.col("health_insurance") + F.col("retirement_contribution") + F.col("other_deductions")
)

display(df_silver_payroll)
```

```python
df_silver_payroll.write.mode("overwrite").saveAsTable("silver.core.payroll")
```
