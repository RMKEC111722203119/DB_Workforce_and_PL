#Core Buisness Logic for General Ledger

## Importing data

```python
import pyspark.sql.functions as F
```

```python
df_silver_ledger = spark.table("silver.transformation.general_ledgers")
```

```python
df_silver_ledger.display()
```

```python
df_silver_ledger.count()
```

## Creating Revenue Column 

```python
df_silver_ledger = df_silver_ledger.withColumn("revenue", F.col("credit_amount") - F.col("debit_amount"))

df_silver_ledger.display()
```

## removing valueless credit and debit

```python
df_silver_ledger = df_silver_ledger.filter((F.col("debit_amount") != 0) | (F.col("credit_amount") != 0))

df_silver_ledger.count()



```

## Filtering Posted rows

```python
df_silver_ledger = df_silver_ledger.filter((F.col("status") == "Posted"))

df_silver_ledger.count()
```

```python
df_silver_ledger.display()
```

##Writing to silver Core schema

```python
df_silver_ledger.write.mode("overwrite").saveAsTable("silver.core.general_ledgers")
```
