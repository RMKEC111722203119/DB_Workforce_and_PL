#Customer Table Cleaning

## importing dependency and data

```python
import pyspark.sql.functions as F
```

```python
df_customer_raw = spark.read.table("bronze.azure_blob_storage.accounts")
```

```python
df_customer_raw.display()
```

## Removing columns inserted by fivetron

```python
df_customer=df_customer_raw.drop('_file','_line','_modified','_fivetran_synced')
```

```python
df_customer.display()
```

## Type Casting columns

```python
df_customer=df_customer.withColumn("account_id", F.col("account_id").cast("int"))
df_customer.display()
```

```python
df_customer.summary().display()
```

## checking for null values

```python
for column in df_customer.columns:
    null_count = df_customer.filter(F.col(column).isNull()).count()
    print(f"{column}: {null_count}")
```

## Checking for duplicates

```python
df_customer.exceptAll(df_customer.dropDuplicates()).show()
```

## Trimming column names

```python
for column in df_customer.columns:
    df_customer=df_customer.withColumn(column,F.trim(F.col(column)))
df_customer.display()
```
