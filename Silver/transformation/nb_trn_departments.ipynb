#Department Table Cleaning

##importing dependency and data

```python
import pyspark.sql.functions as F
```

```python
df_department_raw = spark.read.table("bronze.azure_blob_storage.departments")
```

```python
df_department_raw.display()
```

##Removing columns inserted by fivetron

```python
df_department=df_department_raw.drop('_file','_line','_modified','_fivetran_synced')
```

```python
df_department.display()
```

##Type Casting

```python
df_department =df_department.withColumn("department_id", F.col("department_id").cast("int"))
df_department.display()
```

## Writing to silver

```python
df_department.write.mode("overwrite").saveAsTable("silver.transformation.departments")
```
