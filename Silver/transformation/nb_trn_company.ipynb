#Company Table Cleaning


##importing dependency and data

```python
import pyspark.sql.functions as F
```

```python
df_company_raw = spark.read.table("bronze.azure_blob_storage.company")
```

##Removing columns inserted by fivetron

```python
df_company=df_company_raw.drop('_file','_line','_modified','_fivetran_synced')
```

```python
df_company.display()
```

## Type Casting

```python
df_company  = df_company.withColumn(
    "established_date",
    F.to_date(F.col("established_date"), "dd-MM-yyyy")
)
```

```python
df_company.display()
```

## Writing to Silver

```python
df_company.write.mode('overwrite').saveAsTable("silver.transformation.company")
```
