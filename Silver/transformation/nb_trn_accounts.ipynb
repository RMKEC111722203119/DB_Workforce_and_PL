#Accounts Table Cleaning

## importing dependency and data

```python
import pyspark.sql.functions as F
```

```python
df_accounts_raw = spark.read.table("bronze.azure_blob_storage.accounts")
```

```python
df_accounts_raw.display()
```

## Removing columns inserted by fivetron

```python
df_accounts=df_accounts_raw.drop('_file','_line','_modified','_fivetran_synced')
```

```python
df_accounts.display()
```

## Type Casting columns

```python
df_accounts=df_accounts.withColumn("account_id", F.col("account_id").cast("int"))
df_accounts.display()
```

```python
df_accounts.summary().display()
```

## checking for null values

```python
for column in df_accounts.columns:
    null_count = df_accounts.filter(F.col(column).isNull()).count()
    print(f"{column}: {null_count}")
```

## Checking for duplicates

```python
if df_accounts.count() > df_accounts.dropDuplicates().count():
    df_accounts = df_accounts.dropDuplicates()
display(df_accounts)
```

## Writing to Silver

```python
df_accounts.write.mode("overwrite").saveAsTable("silver.transformation.accounts")
```
