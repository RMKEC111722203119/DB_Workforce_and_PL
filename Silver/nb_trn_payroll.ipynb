```python
import pyspark.sql.functions as F
from pyspark.sql.functions import *
```

```python
df_payroll_raw = spark.table("bronze.azure_blob_storage.payroll")
```

```python
df_payroll_raw.display()
```

## Removing columns inserted by fivetron

```python
df_payroll=df_payroll_raw.drop('_file','_line','_modified','_fivetran_synced')
```

```python
df_payroll.display()
```

## Trimming column Names

```python
for column in df_payroll.columns:
    df_payroll=df_payroll.withColumn(column,F.trim(F.col(column)))
df_payroll.display()
```

```python

df_payroll = df_payroll.withColumn("pay_period_start", to_date(col("pay_period_start"), "dd-MM-yyyy HH:mm")) \
    .withColumn("pay_period_end", to_date(col("pay_period_end"), "dd-MM-yyyy HH:mm"))\
    .withColumn("pay_date", to_date(col("pay_date"), "dd-MM-yyyy HH:mm")) 
```

```python
df_payroll.display()
```

```python
df_payroll=df_payroll.withColumn("gross_salary",col("gross_salary").cast("double"))
df_payroll=df_payroll.withColumn("bonus",col("bonus").cast("double"))
df_payroll=df_payroll.withColumn("overtime_pay",col("overtime_pay").cast("double"))
df_payroll=df_payroll.withColumn("commission",col("commission").cast("double"))
df_payroll=df_payroll.withColumn("allowances",col("allowances").cast("double"))
df_payroll=df_payroll.withColumn("tax_deduction",col("tax_deduction").cast("double"))
```

```python
df_payroll.display()
```

```python
df_payroll=df_payroll.withColumn("social_security",col("social_security").cast("double"))
df_payroll=df_payroll.withColumn("health_insurance",col("health_insurance").cast("double"))
df_payroll=df_payroll.withColumn("retirement_contribution",col("retirement_contribution").cast("double"))
df_payroll=df_payroll.withColumn("other_deductions",col("other_deductions").cast("double"))
df_payroll=df_payroll.withColumn("allowances",col("allowances").cast("double"))
df_payroll=df_payroll.withColumn("net_salary",col("net_salary").cast("double"))
```

```python
df_payroll=df_payroll.withColumn("payroll_id",col("payroll_id").cast("int"))
df_payroll=df_payroll.withColumn("employee_id",col("employee_id").cast("int"))
df_payroll=df_payroll.withColumn("company_id",col("company_id").cast("int"))
df_payroll=df_payroll.withColumn("department_id",col("department_id").cast("int"))
```

```python
df_payroll.display()
```
