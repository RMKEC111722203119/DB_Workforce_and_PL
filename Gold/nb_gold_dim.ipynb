# Dim Table Conversion

### Department Table

```python
df_dim_department = spark.table("silver.core.departments") \
    .select(
        "department_id",
        "department_name",
        "department_code",
        "cost_center"
    )

df_dim_department.write.mode("overwrite").saveAsTable("gold.dim.dim_departments")
```

### Company Table

```python
df_dim_company = spark.table("silver.core.company") \
    .select(
        "company_id",
        "company_name",
        "industry",
        "country",
        "established_date"
    )

df_dim_company.write.mode("overwrite").saveAsTable("gold.dim.dim_company")
```

### Employee Table

```python
df_dim_employee = spark.table("silver.core.employee") \
    .select(
        "employee_id",
        "employee_code",
        "first_name",
        "last_name",
        "email",
        "department_id",
        "company_id",
        "position",
        "hire_date",
        "termination_date"
    )

df_dim_employee.write.mode("overwrite").saveAsTable("gold.dim.dim_employee")
```

### Account Table

```python
df_dim_account = spark.table("silver.core.accounts") \
    .select(
        "account_id",
        "account_code",
        "account_name",
        "account_type",
        "category"
    )

df_dim_account.write.mode("overwrite").saveAsTable("gold.dim.dim_account")
```
