#KPI Cubes

### Cube 1 Monthly review cube

```python
%sql
CREATE OR REPLACE TABLE gold.cube.cube_monthly_finance AS

SELECT
    r.year,
    r.month,
    c.company_id,

    r.total_revenue,
    c.total_cost_of_sales,
    g.gross_profit,
    g.gross_profit_margin,
    n.net_profit,
    p.payroll_to_revenue_ratio,
    e.total_expense

FROM gold.kpi.kpi_1_monthly_revenue r

LEFT JOIN gold.kpi.kpi_2_cost_of_sales c
    ON r.year = c.year AND r.month = c.month

LEFT JOIN gold.kpi.kpi_3_gross_profit_margin g
    ON r.year = g.year AND r.month = g.month

LEFT JOIN gold.kpi.kpi_6_net_profit n
    ON r.year = n.year AND r.month = n.month

LEFT JOIN gold.kpi.kpi_10_payroll_to_revenue p
    ON r.year = p.year AND r.month = p.month

LEFT JOIN gold.kpi.kpi_4_operating_expense_breakdown e
    ON r.year = e.year 
   AND r.month = e.month 
   AND c.company_id = e.company_id;
```

```python
%sql
select * from gold.cube.cube_monthly_finance
```

### Cube 2 department review cube

```python
%sql

CREATE OR REPLACE TABLE gold.cube.cube_department AS

SELECT
    d.department_id,
    d.department_name,

    k8.total_department_cost,
    k9.headcount,
    k7.total_overtime,
    k7.total_bonus,
    k7.overtime_to_salary_ratio

FROM gold.kpi.kpi_8_cost_per_department k8

LEFT JOIN gold.kpi.kpi_9_headcount k9
    ON k8.department_id = k9.department_id

LEFT JOIN gold.kpi.kpi_7_overtime_bonus k7
    ON k8.department_id = k7.department_id

LEFT JOIN gold.dim.dim_departments d  
    ON k8.department_id = d.department_id;
```

```python
%sql
select * from gold.cube.cube_department
```

### Position Cube

```python
%sql
CREATE OR REPLACE TABLE gold.cube.cube_position AS

SELECT
    company_id,
    position,
    avg_compensation
FROM gold.kpi.kpi_5_avg_compensation;
```

```python
%sql
select * from  gold.cube.cube_position
```
