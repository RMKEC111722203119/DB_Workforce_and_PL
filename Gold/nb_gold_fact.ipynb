#Fact Table Conversion

### General ledger Table

```python
df_fact_ledger = spark.table("silver.core.general_ledgers") \
    .select(
        "gl_id",
        "posting_date",
        "company_id",
        "account_id",
        "department_id",
        "transaction_type",
        "revenue" 
    )

df_fact_ledger.write.mode("overwrite").saveAsTable("gold.fact.fact_general_ledgerse")
```

### Payroll Table

```python
fact_payroll = spark.table("silver.core.payroll") \
    .select(
        "payroll_id",
        "employee_id",
        "company_id",
        "department_id",
        "pay_date",
        "gross_salary",
        "bonus",
        "overtime_pay",
        "total_compensation",
        "total_deduction"
    )

fact_payroll.write.mode("overwrite").saveAsTable("gold.fact.fact_payroll")
```
